# files and dir

# dirでのやり取り

## cwd

```python
import os
 
cwd = os.getcwd()
cwd
```

## mkdir, cd, ls

```python
new_dir = 'test_dir'
os.mkdir(new_dir)
os.chdir(new_dir)
os.getcwd()
os.listdir()
```

## globでfull_path表示

```python
import glob

files = glob.glob(cwd+'/'+ new_dir+'/*')
for file in files:
    print(file)
```

## 単純な読み書き

```python
data = ['x', 'y']
text = "\n".join(data)
with open("tmp.txt", "w") as f:
    f.write(text)
```

```python
with open("tmp.txt", "r") as f:
    lines = f.readlines()
    print(lines)
```

```python
for line in lines:
    print(line.rstrip())
```

## sprintfからの出力

```python
data = [['x', 0, 'y', 0],
        ['x', 1, 'y', 1]]

with open("tmp.txt", "w") as f:
    for dd in data:
        print(dd)
        string = '%4s:%3d, %4s:%3d\n' % tuple(dd)
        print(string)
        f.writelines(string)
```

```python
data = [['x', 0, 'y', 0],
        ['x', 1, 'y', 1]]

f = open("tmp.txt", "w")
for dd in data:
    print(dd)
    string = '%4s:%3d, %4s:%3d\n' % tuple(dd)
    print(string)
    f.writelines(string)

f.close()
```

```python
file = open("tmp.txt", "r")
lines = file.readlines()
for line in lines:
    print(line.rstrip())
```
